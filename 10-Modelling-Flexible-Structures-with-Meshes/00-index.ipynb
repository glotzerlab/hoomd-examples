# Modelling Flexible Active Superstructures ("Flexicles")

The motion of biological cells like amoeba are fascinating because they can autonomously adapt to their local environments morphologically and dynamically. Dr. Philipp Schoenhoefer has developed a computational model for a 3D morphologically flexible particle that mimics the desired behavior of these biological systems. The framework models amoeba-like structures as collections of self-propelled particles confined within a triangulated mesh membrane. We refer to the resulting superstructure as a [“flexicle”](cite_philipp_here). One of the most exciting aspects of the flexicle model is the vast design space it offers; variations in particle size, shape, membrane rigidity, and interaction potentials can all be systematically explored. This breadth of design parameters opens numerous avenues in the realm soft robotics and biomedicine.
