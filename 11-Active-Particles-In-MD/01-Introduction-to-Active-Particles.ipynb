## Introduction to Active Particles

**Active particles** are particles that can consume energy from the environment and convert this into directed motion. In experiments, there are many different types of active particles with different mechanisms for this realized self-propulsion.

For example, a polystyrene bead coated in platinum on one size (ie. an example of a simple **Janus particle**, a particle with 2 of more regions of distinct physical properties) will be propelled forward if submerged in a solution containing hydrogen peroxide. This occurs because the platinum catalyzes a reaction that decomposes the hydrogen peroxide into oxygen gas and water. The localized chemical reaction propels the particle forward. In this example, thermal and chemical energy is being consumed to generate the kinetic motion of the (now active) janus particle.

In the previous hoomd tutorials, most of the systems we've looked at have been good models to investigate **equilibrium properties** where the steady state is independent of kinetic parameters. When studying **active matter**, however, we are no longer considering equilibrium processes and shift our attention to **non-equilibrium dynamics**.
