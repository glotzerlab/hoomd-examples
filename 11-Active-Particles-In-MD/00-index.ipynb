# Modelling Active Particles in MD

**Active particles** are particles that can consume energy from the environment and convert this into directed motion. In experiments, there are many different types of active particles with different mechanisms for this realized self-propulsion. The majority of HOOMD tutorials until now focus on simulating systems in **equilibium**. However, when applying an active force in HOOMD, we can simulate systems of active particles and investigate **non-equilibrium dynamics** such as **motility induced phase separation (MIPS)** among other cool effects.  

**Prerequisites:**

- This tutorial assumes you have some familiarity with the Python programming language.
- This tutorials assume basic understanding of the fundamentals of HOOMD-Blue.
- This tutorial uses an example with rod-shaped self-propelled particles which are made with Rigid Bodies in MD. Please refer to the [Modelling Rigid Bodies](../06-Modelling-Rigid-Bodies/00-index.ipynb) tutorial for more details as to how this is done.
- This tutorial uses walls as an example way to confine your active bodies. Please refer to the [Modelling Rigid Bodies](../06-Modelling-Rigid-Bodies/00-index.ipynb) tutorial for more details as to how this is done.
